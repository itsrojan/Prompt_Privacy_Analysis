```python
import pandas as pd
import numpy as np

file_name = 'ChatGPT_Prompt_Response_Pairs_Dataset_with_Topic.xlsx'

# Read sheet 1 (main data) and sheet 2 (topic counts)
df_main = pd.read_excel(file_name, sheet_name=0)
df_sampling = pd.read_excel(file_name, sheet_name=1) # sheet_name=1 refers to Sheet 2

# Exclude the 2 last row from stratification calculations ('Total' row and blank row)
df_sampling = df_sampling.iloc[:-2]

# Calculate the number of samples to select per topic
total_samples = df_sampling['Count of Topic'].sum()
print(f"Total samples: {total_samples}")
sample_size = 50
df_sampling['Proportion'] = df_sampling['Count of Topic'] / total_samples
df_sampling['Samples_to_Select'] = (df_sampling['Proportion'] * sample_size).round().astype(int)

# Adjust in case rounding causes the sum to differ from sample_size
difference = sample_size - df_sampling['Samples_to_Select'].sum()
if difference != 0:
    remainders = (df_sampling['Proportion'] * sample_size) - df_sampling['Samples_to_Select']
    adjust_indices = remainders.abs().sort_values(ascending=False).index[:abs(difference)]
    df_sampling.loc[adjust_indices, 'Samples_to_Select'] += int(np.sign(difference))

df_display = df_sampling[['Topic List', 'Samples_to_Select', 'Proportion', 'Count of Topic']].copy()
df_display['Proportion'] = (df_display['Proportion'] * 100).round(2).astype(str) + '%'
df_display
```
